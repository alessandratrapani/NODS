```python
result_path = "/g100_work/EIRI_E_POLIMI/no_paper/NODS/results/grid_search_new/"
#result_path = "/g100_scratch/userexternal/csartor1/results/without_NO/grid_search/"
with open("/g100_work/EIRI_E_POLIMI/no_paper/NODS/network_configuration.json", "r") as json_file:
    net_config = json.load(json_file)
```

```python
CS_burst_dur = net_config["devices"]["CS"]["parameters"]["burst_dur"]
CS_start_first = float(net_config["devices"]["CS"]["parameters"]["start_first"])
between_start = net_config["devices"]["CS"]["parameters"]["between_start"]
n_trials = net_config["devices"]["CS"]["parameters"]["n_trials"]
US_start_first = float(net_config["devices"]["US"]["parameters"]["start_first"])
num_sim = 100
trials_to_eval = 3
trial_length = between_start
```

```python
minus = [12,11,10,9,8,7,6,5,4,3]
plus = [10,9,8,7,6,5,4,3]
```

```python
grid_search = np.nan*np.ones((len(minus),len(plus)))
grid_cr = np.nan*np.ones((len(minus),len(plus)))
grid_frequency = np.nan*np.ones((len(minus),len(plus)))
sim_baseline = np.zeros((len(minus),len(plus),num_sim*trials_to_eval))
sim_cr = np.zeros((len(minus),len(plus),num_sim*trials_to_eval))
sim_frq = np.zeros((num_sim,len(minus),len(plus)))


for i,m in enumerate(minus):
    for j,p in enumerate(plus):
        folder_path = result_path + f"minus{m}_plus{p}/"
        if not os.path.exists(folder_path):
            continue
        for k in range (0,num_sim):
            file_path = folder_path+ f"sim{k}/"
            
            df_sdf_change = pd.read_csv(os.path.join(file_path,'sdf_grid.csv'))
            sdf_baseline = df_sdf_change.iloc[:,0].values
            sdf_cr = df_sdf_change.iloc[:,1].values
        
            sdf_change_baseline = sdf_baseline[3:] - np.median(sdf_baseline[1:3]) #sdf_baseline[1] #
            sdf_change_cr = sdf_cr[3:] - np.median(sdf_cr[1:3]) #sdf_cr[1] #
        
            sim_baseline[i,j,k*trials_to_eval:(k+1)*trials_to_eval] = sdf_change_baseline[-trials_to_eval:]
            sim_cr[i,j,k*trials_to_eval:(k+1)*trials_to_eval] = sdf_change_cr[-trials_to_eval:]
            sim_frq[k,i,j] = sdf_cr[-1]
                                  
        # learning in respect to the baseline
        grid_search[i,j] = (np.median(sim_baseline[i,j]) - np.median(sim_cr[i,j]))
        grid_cr[i,j]= -1*np.median(sim_cr[i,j])
        
        # pc frequency
grid_frequency = np.median(sim_frq, axis = 0)
```
